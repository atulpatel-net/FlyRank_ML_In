## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### ML Task

The primary ML task for this project is **scoring and ranking**.

Instead of predicting a simple yes/no outcome, the goal is to assign each content page a refresh priority score and rank the pages from highest to lowest priority. This helps content editors focus on the pages that are most likely to benefit from a review, making the recommendations more useful than a binary classification.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target / Proxy

The target for this project is a **refresh priority score** for each content page. Since there is no direct label that tells us which pages should be refreshed, the priority will be estimated using observable search and content signals such as trend direction, impressions, CTR, content age, and other available features. This score will be used to rank pages for review rather than make absolute predictions.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

The primary success metric for this project is **Precision@20**. Since the goal is to rank pages for review, the most important question is whether the top recommendations are actually useful. A higher Precision@20 means that more of the top 20 recommended pages are genuinely good candidates for content review, making the recommendations more valuable for content editors.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML is a better approach

A fixed rule usually depends on one or two conditions, such as low impressions or old content. However, whether a page should be reviewed often depends on multiple factors working together, including impressions, CTR, average position, content age, engagement, and recent performance trends. A machine learning model can combine these signals to produce a more reliable refresh priority score and rank pages more effectively than a single hand-written rule.
