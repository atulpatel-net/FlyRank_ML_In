## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("content_refresh_anonymized.csv")

# Supporting statistics
total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
old_pages = (df["content_age_days"] > 365).sum()

print(f"Total pages: {total_pages}")
print(f"Pages with a downward trend: {declining_pages}")
print(f"Pages older than one year: {old_pages}")
```

### Observation

The starter dataset contains **30,000** content pages, making manual review impractical. Among them, **16,262** pages are showing a downward trend, and **6,360** pages are more than one year old. These numbers indicate that many pages could require attention, making a data-driven approach to prioritize content reviews both useful and scalable.
